## 🟢 A1 — Verbo correcto
Para cada acción, decí qué verbo HTTP corresponde y por qué:

(a) Ver la lista de libros - GET

(b) Agregar un libro nuevo - POST

(c) Borrar un libro - DELETE

(d) Cambiarle el precio a un libro - PACHT

(e) Reemplazar un libro entero por otro  - PUT

(f) Cambiarle solo la disponibilidad a un libro sin tocar el resto - PACHT

**Diferencia entre (e) y (f):**
PUT, reemplaza el objeto entero (si no mandás todos los campos, los que faltan se pierden o quedan mal); PATCH toca únicamente el/los campos que le pasás, y el resto del recurso queda intacto.

## 🟢 A2 — Leer códigos de estado
¿Qué significa cada uno y cuándo lo devolverías? Dá un ejemplo concreto en el dominio de libros.

- `200` → "ok" > libro encontrado.
- `201` → "Created" > libro creado.
- `204` → "not contend" > libro borrado con exito.
- `400` → "bad request" > pedido mal hecho,ej: tal libro no existe.
- `404` → "not found" > libro no encontrado
- `405` → "method not allowed" > metodo no permitido para los libros
- `422` → "unprocessable entity" > el pedido esta mal formado.
- `500` → "internal server error" > la biblioteca esta fallando

🟢 A3 — Familias de códigos:

2xx → Salió bien, el pedido se procesó como se esperaba.

3xx → Redirección: lo que pediste está en otro lado, hay que ir a buscarlo ahí.

4xx → El cliente se equivocó en algo (pidió mal, o pidió algo que no existe).

5xx → El servidor se rompió (no es culpa de quien hizo el pedido).

Con esta lógica, un código nuevo se puede clasificar sin memorizarlo:

301 → empieza con 3 → es una redirección (el recurso se movió a otra URL).

403 → empieza con 4 → error del cliente (en este caso, no tiene permiso para acceder, aunque el recurso exista).

## 🟢 A6 — Headers, lo mínimo
¿Para qué sirve el header `Content-Type: application/json`?
el header es como la extencion de cada archivo(.py,.jpg.txt)sirve para avisarle al server que el archivo esta en formato JSON.

¿Qué pasaría si un cliente manda un body JSON sin ese header?

Sin ese header, algunas APIs podrían no entender correctamente el body. O podrían rechazar la petición porque no saben cómo interpretar los datos.
En JSONPlaceholder muchas cosas funcionan aunque no seas tan estricto, porque está hecho para practicar. Pero en APIs reales, estos detalles sí importan.
