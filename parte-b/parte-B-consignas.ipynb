# Parte B — Tu primera API (FastAPI), dominio libros

## 🟢 B1 — Endpoint que saluda
Levantá FastAPI con un solo endpoint `GET /` que devuelva `{"mensaje": "hola"}`. Probalo en `/docs`.

**Listo cuando:** ves la respuesta en el navegador y en `/docs`.



## 🟡 B2 — Listar (GET)
Agregá `GET /libros` que devuelva una lista de libros (fija, en una variable, con al menos 3 libros precargados).

**Listo cuando:** `/docs` te muestra la lista al ejecutarlo.

## 🟡 B3 — Crear (POST)
Agregá `POST /libros` que reciba un libro (con su modelo de datos: `titulo`, `paginas`) y lo agregue a la lista. Después listá con el GET y verificá que aparezca.

**Listo cuando:** cargás por POST y el dato aparece en el GET.

## 🟡 B4 — Provocá el 422
Mandá un POST **sin** el campo `paginas`, otro con `paginas: "muchas"` (texto en vez de número), y otro con un campo **de más** que el modelo no espera (ej. `"editorial": "x"`). Leé el cuerpo del error en los dos primeros casos, y fijate qué pasa con el tercero.

**Listo cuando:** entendés *qué campo* y *qué tipo* esperaba la API leyéndolo del 422, y sabés si FastAPI rechaza o ignora un campo de más (probalo, no lo asumas).

## 🟡 B5 — Buscar uno (GET con parámetro)
Agregá `GET /libros/{titulo}` que devuelva un libro por su título, o `404` si no existe.

**Listo cuando:** un título que existe devuelve el libro y uno que no, da 404.

## 🟡 B6 — Actualizar (PUT)
Agregá `PUT /libros/{titulo}` que reemplace el libro entero por uno nuevo. Si el título no existe, devolvé `404`.

**Listo cuando:** un PUT sobre un título existente reemplaza todos sus campos, y sobre uno inexistente da 404 en vez de crearlo silenciosamente.

## 🟡 B7 — Borrar (DELETE)
Agregá `DELETE /libros/{titulo}`. Decidí qué código de estado devolver cuando sale bien (pista: `204` es una opción válida, ¿por qué no lleva body?) y qué devolver si no existe.

**Listo cuando:** después de borrar, un GET a ese título da 404, y elegiste el código de éxito con criterio (no "porque sí").

## 🔴 B8 — Validación propia
Hacé que la API rechace un libro con `paginas` menor o igual a 0. Decidí qué código de estado devolver y por qué.

**Listo cuando:** un libro con 0 páginas no entra y la API explica el motivo.

## 🔴 B9 — Segundo recurso
Sumá un recurso totalmente distinto, ej. `GET/POST /autores`. Reutilizá lo que aprendiste.

**Listo cuando:** la API maneja dos recursos sin pisarse.

## 🔴 B10 — Modelo anidado
Hacé que `Libro` tenga un campo `editorial` que sea a su vez un objeto con `nombre` y `pais` (dos modelos Pydantic, uno adentro del otro). Cargá un libro con editorial desde `/docs` y mirá cómo se ve el formulario que FastAPI genera solo.

**Listo cuando:** creás un libro con editorial anidada y el GET te la devuelve completa, con la misma estructura anidada.

## 🔴 B11 — Filtrar con query params
Agregá a `GET /libros` un parámetro opcional `paginas_min` (ej. `GET /libros?paginas_min=300`) que devuelva solo los libros con al menos esa cantidad de páginas. Si no se manda el parámetro, devolvé todos.

**Listo cuando:** `/libros` sin parámetro trae todo, y `/libros?paginas_min=300` filtra, probado desde `/docs`.

## 🔴 B12 — Valores por defecto y `response_model`
Agregá al modelo `Libro` un campo `disponible: bool = True` (valor por defecto). Investigá qué es el parámetro `response_model` de FastAPI y para qué sirve (pista: pensá qué pasa si querés devolver un libro pero **sin** un campo interno, como un precio de costo).

**Listo cuando:** un POST sin mandar `disponible` lo asume `True`, y podés explicar en una frase para qué sirve `response_model`.
